```python
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow .keras.utils  import pad_sequences
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense, SimpleRNN, Embedding, Flatten
```

```python
(X_train, y_train), (X_test, y_test) = imdb.load_data(num_words=10000)
```

```python
X_train = pad_sequences(X_train, padding='post', maxlen=50)
X_test = pad_sequences(X_test, padding='post', maxlen=50)
```

```python
X_train.shape
```

```python
model = Sequential()
model.add(Embedding(input_dim=10000, output_dim=2, input_length=50))
model.add(SimpleRNN(32, return_sequences=False))
model.add(Dense(1, activation='sigmoid'))
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['acc'])
history = model.fit(X_train, y_train, epochs=10, validation_data=(X_test, y_test))
```

```python
# Get the word index uesd by the IMDB dataset
word_index = imdb.get_word_index()

#Adjust incides (keras reserves 0-3)
word_index = {k: (v + 3) for k, v in word_index.items()}
word_index['<PAD>'] = 0
word_index['<START>'] = 1
word_index['<UNK>'] = 2
word_index['<UNUSED>'] = 3
```

```python
def simple_encode(text):
  words = text.lower().split()
  encoded = [1]
  for word in words :
    encoded.append(word_index.get(word, 2))
  return pad_sequences ([encoded], maxlen=50, padding='post')
```

```python
#Your custom sentence
text = "the news was bad"

#Encode and Predict
encoded = simple_encode(text)
pred = model.predict(encoded)

print("Sentiment score:", pred[0][0])
print("prediction:"," positive " if pred[0][0] >= 0.5 else "Negative")
```
